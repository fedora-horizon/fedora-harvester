```python
# create organization 
def create_organization(rows: list) :
    header["Content-Type"] = "application/json"
    for row in rows :
        payload = {
            "name": row['owner_org'],
            "title": row['title'],
            "description": row['description']
        }
        http.post(path="/api/3/action/organization_create", json=payload, headers=header)
```
