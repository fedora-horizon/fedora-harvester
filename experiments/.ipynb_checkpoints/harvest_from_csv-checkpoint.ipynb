```python
import requests

CKAN_URL = os.getenv("CKAN_URL")
API_KEY = os.getenv("CKAN_API_KEY")

headers = {
    "Authorization": API_KEY,
    "Content-Type": "application/json"
}

# 1. Get all datasets
resp = requests.post(
    f"{CKAN_URL}/api/3/action/package_list",
    headers=headers,
    json={},
    verify=False
)
datasets = resp.json()["result"]

# 2. Purge each dataset
for ds in datasets:
    r = requests.post(
        f"{CKAN_URL}/api/3/action/dataset_purge",
        headers=headers,
        json={"id": ds},
        verify=False
    )
    print(ds, r.status_code, r.json())
```
