```python
import os
import requests
def fetch_orgs(base_url="https://48.220.49.10", api_key=None):
    headers = {}
    if api_key:
        headers["Authorization"] = os.getenv("CKAN_API_KEY")
    session = requests.Session()
    r = session.get(f"{base_url}/api/3/action/organization_list",
                     headers=headers,
                     verify=False)
    return r.json()
```
